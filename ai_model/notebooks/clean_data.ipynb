## Feature Engineering


```python
def calculate_soil_impact(crop_name):
    name = str(crop_name).lower()
    nitrogen_fixers = [
        "soybean",
        "clover",
        "alfalfa",
        "pea",
        "bean",
        "lentil",
        "chickpea",
        "vetch",
        "lupin",
    ]
    heavy_feeders = [
        "maize",
        "corn",
        "cotton",
        "sugarcane",
        "wheat",
        "sunflower",
        "potato",
        "tobacco",
        "sorghum",
    ]

    if any(fixer in name for fixer in nitrogen_fixers):
        return 1
    elif any(feeder in name for feeder in heavy_feeders):
        return -1
    else:
        return 0
```

```python
df["Soil_Impact"] = df["COMNAME"].apply(calculate_soil_impact)
```
